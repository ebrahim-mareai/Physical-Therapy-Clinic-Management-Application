```python
# 1. imports and colors
COLOR_BG = "#F4F7F6"
COLOR_PRIMARY = "#2A7F9E"
COLOR_TEXT = "#2C3E50"
COLOR_LEFT_BG = "#E1EBEE"
COLOR_RIGHT_BG = "#FFFFFF"
COLOR_INPUT_BG = "#FFFFFF"
COLOR_BTN_TEXT = "#FFFFFF"
```

### Building the Main Split-Screen Layout
In this cell, We divide the screen into two sides: 
1. **Left Side:** A control panel where we will later place all the push buttons.
2. **Right Side:** A flexible results area where we will later display the patient queue and names.

```python
# 2. main split screen
def build_main_ui(root):
    # main container frame (medical bg)
    main_frame = tk.Frame(root, bg=COLOR_BG)
    main_frame.pack(fill=tk.BOTH, expand=True)
    # left panel (fixed 300px)
    left_frame = tk.Frame(main_frame, bg=COLOR_LEFT_BG, width=300, relief=tk.RAISED, borderwidth=1)
    left_frame.pack(side=tk.LEFT, fill=tk.Y, padx=10, pady=10)
    left_frame.pack_propagate(False)
    tk.Label(left_frame, text="Control Panel", font=("Arial", 14, "bold"), bg=COLOR_LEFT_BG, fg=COLOR_PRIMARY).pack(pady=20)
    # right panel (results, white bg)
    right_frame = tk.Frame(main_frame, bg=COLOR_RIGHT_BG, relief=tk.SUNKEN, borderwidth=1)
    right_frame.pack(side=tk.RIGHT, fill=tk.BOTH, expand=True, padx=10, pady=10)
    tk.Label(right_frame, text="Patient Queue & Results", font=("Arial", 14, "bold"), bg=COLOR_RIGHT_BG, fg=COLOR_PRIMARY).pack(pady=20)
    return left_frame, right_frame
```

### Building the Nurse's Input Form
###### In this cell, we create a function that adds the input fields to the left side of screen. 
###### The nurse will use these text boxes to type the patient's Name, Phone Number, and Age, and then click the button to add them to the queue. 

```python
def build_nurse_form(left_frame):
    # input section title (medical teal, bold)
    title = tk.Label(left_frame, text="New Patient Details", font=("Arial", 14, "bold"), bg=COLOR_LEFT_BG, fg=COLOR_PRIMARY)
    title.pack(pady=(0, 15))
    
    # full name label
    tk.Label(left_frame, text="Full Name:", font=("Arial", 11), bg=COLOR_LEFT_BG, fg=COLOR_TEXT).pack(anchor=tk.W, padx=15)
    
    # name entry box (fills width)
    entry_name = tk.Entry(left_frame, font=("Arial", 12), bg=COLOR_INPUT_BG, fg=COLOR_TEXT, relief=tk.SOLID, borderwidth=1)
    entry_name.pack(fill=tk.X, padx=15, pady=(0, 10))
    
    # phone number label
    tk.Label(left_frame, text="Phone Number:", font=("Arial", 11), bg=COLOR_LEFT_BG, fg=COLOR_TEXT).pack(anchor=tk.W, padx=15)
    
    # phone entry box
    entry_phone = tk.Entry(left_frame, font=("Arial", 12), bg=COLOR_INPUT_BG, fg=COLOR_TEXT, relief=tk.SOLID, borderwidth=1)
    entry_phone.pack(fill=tk.X, padx=15, pady=(0, 10))
    
    # patient age label
    tk.Label(left_frame, text="Patient Age:", font=("Arial", 11), bg=COLOR_LEFT_BG, fg=COLOR_TEXT).pack(anchor=tk.W, padx=15)
    
    # age entry box
    entry_age = tk.Entry(left_frame, font=("Arial", 12), bg=COLOR_INPUT_BG, fg=COLOR_TEXT, relief=tk.SOLID, borderwidth=1)
    entry_age.pack(fill=tk.X, padx=15, pady=(0, 15))
    
    # payment status label (left-aligned)
    tk.Label(left_frame, text="Payment Status:", font=("Arial", 11), bg=COLOR_LEFT_BG, fg=COLOR_TEXT).pack(anchor=tk.W, padx=15)
    
    # payment dropdown (readonly)
    combo_payment = ttk.Combobox(left_frame, values=["Unpaid", "Paid", "Partial"], state="readonly", font=("Arial", 12))
    combo_payment.current(0)
    combo_payment.pack(fill=tk.X, padx=15, pady=(0, 15))
    
    # add to queue button
    btn_add = tk.Button(left_frame, text="Add to Queue", font=("Arial", 12, "bold"), bg=COLOR_PRIMARY, fg=COLOR_BTN_TEXT, activebackground="#1F6077", activeforeground=COLOR_BTN_TEXT, relief=tk.FLAT, cursor="hand2")
    btn_add.pack(fill=tk.X, padx=15, ipady=5)
    
    # return all 5 widgets for business logic
    return entry_name, entry_phone, entry_age, combo_payment, btn_add
```

# 5. patient queue table


```python
def build_patient_queue(right_frame):
    # waiting queue title
    tk.Label(right_frame, text="Waiting Patients Queue", font=("Arial", 14, "bold"), bg=COLOR_RIGHT_BG, fg=COLOR_PRIMARY).pack(pady=(0, 10))
    # container for table + scrollbar
    tree_frame = tk.Frame(right_frame, bg=COLOR_RIGHT_BG)
    tree_frame.pack(fill=tk.BOTH, expand=True, padx=15, pady=10)
    # queue table (4 columns)
    tree = ttk.Treeview(tree_frame, columns=("ID", "Name", "Age", "Phone"), show="headings")
    tree.heading("ID", text="ID")
    tree.column("ID", width=50, anchor=tk.CENTER)
    tree.heading("Name", text="Patient Name")
    tree.column("Name", width=200, anchor=tk.W)
    tree.heading("Age", text="Age")
    tree.column("Age", width=60, anchor=tk.CENTER)
    tree.heading("Phone", text="Phone Number")
    tree.column("Phone", width=150, anchor=tk.W)
    # vertical scrollbar
    scrollbar = ttk.Scrollbar(tree_frame, orient=tk.VERTICAL, command=tree.yview)
    tree.configure(yscrollcommand=scrollbar.set)
    tree.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)
    scrollbar.pack(side=tk.RIGHT, fill=tk.Y)
    return tree
```

# 10. recovery plan window


```python
# build recovery plan window
def build_recovery_plan_window(parent, patient_name):
    # pop-up window
    plan_window = tk.Toplevel(parent)
    plan_window.title(f"Recovery Plan - {patient_name}")
    plan_window.geometry("600x550")
    plan_window.configure(bg=COLOR_BG)
    # title
    tk.Label(plan_window, text=f"Creating Plan for: {patient_name}", font=("Arial", 16, "bold"), bg=COLOR_BG, fg=COLOR_PRIMARY).pack(pady=15)
    # input container (grid layout)
    input_frame = tk.Frame(plan_window, bg=COLOR_BG)
    input_frame.pack(pady=10, padx=30, fill=tk.X)
    # step description input
    tk.Label(input_frame, text="Step Description (e.g., Massage):", font=("Arial", 11), bg=COLOR_BG, fg=COLOR_TEXT).grid(row=0, column=0, sticky=tk.W, pady=8)
    entry_desc = tk.Entry(input_frame, font=("Arial", 12), bg=COLOR_INPUT_BG, fg=COLOR_TEXT, relief=tk.SOLID, borderwidth=1)
    entry_desc.grid(row=0, column=1, padx=10, pady=8, sticky=tk.EW)
    # device name input
    tk.Label(input_frame, text="Device Name (e.g., Relax Device):", font=("Arial", 11), bg=COLOR_BG, fg=COLOR_TEXT).grid(row=1, column=0, sticky=tk.W, pady=8)
    entry_device = tk.Entry(input_frame, font=("Arial", 12), bg=COLOR_INPUT_BG, fg=COLOR_TEXT, relief=tk.SOLID, borderwidth=1)
    entry_device.grid(row=1, column=1, padx=10, pady=8, sticky=tk.EW)
    # duration input
    tk.Label(input_frame, text="Duration (minutes):", font=("Arial", 11), bg=COLOR_BG, fg=COLOR_TEXT).grid(row=2, column=0, sticky=tk.W, pady=8)
    entry_duration = tk.Entry(input_frame, font=("Arial", 12), bg=COLOR_INPUT_BG, fg=COLOR_TEXT, relief=tk.SOLID, borderwidth=1)
    entry_duration.grid(row=2, column=1, padx=10, pady=8, sticky=tk.EW)
    input_frame.columnconfigure(1, weight=1)
    # add step button
    btn_add_step = tk.Button(plan_window, text="Add Step to List", font=("Arial", 12, "bold"), bg=COLOR_PRIMARY, fg=COLOR_BTN_TEXT, relief=tk.FLAT, cursor="hand2")
    btn_add_step.pack(pady=10)
    # steps table container
    tree_frame = tk.Frame(plan_window, bg=COLOR_BG)
    tree_frame.pack(fill=tk.BOTH, expand=True, padx=30, pady=10)
    # steps table
    tree_steps = ttk.Treeview(tree_frame, columns=("Order", "Description", "Device", "Duration"), show="headings")
    tree_steps.heading("Order", text="Step #")
    tree_steps.column("Order", width=50, anchor=tk.CENTER)
    tree_steps.heading("Description", text="Description")
    tree_steps.column("Description", width=200, anchor=tk.W)
    tree_steps.heading("Device", text="Device Used")
    tree_steps.column("Device", width=150, anchor=tk.W)
    tree_steps.heading("Duration", text="Mins")
    tree_steps.column("Duration", width=60, anchor=tk.CENTER)
    tree_steps.pack(fill=tk.BOTH, expand=True)
    # save plan button (green)
    btn_save_plan = tk.Button(plan_window, text="Save Plan to Database", font=("Arial", 12, "bold"), bg="#28A745", fg=COLOR_BTN_TEXT, relief=tk.FLAT, cursor="hand2")
    btn_save_plan.pack(pady=15, ipady=5, ipadx=20)
    # return widgets
    return {
        "window": plan_window,
        "entry_desc": entry_desc,
        "entry_device": entry_device,
        "entry_duration": entry_duration,
        "btn_add_step": btn_add_step,
        "tree_steps": tree_steps,
        "btn_save_plan": btn_save_plan
    }
```

```python

# build patient history window
def build_patient_history_window(parent, patient_name=None, patient_id=None):
    window = tk.Toplevel(parent)
    window.title("Visitor History")
    window.geometry("750x450")

    # window title
    tk.Label(
        window,
        text="Patient Visitor History",
        font=("Arial", 15, "bold")
    ).pack(pady=10)

    # table columns
    columns = (
        "ID",
        "Name",
        "Phone",
        "Age",
        "Payment",
        "Visit Date"
    )

    tree = ttk.Treeview(
        window,
        columns=columns,
        show="headings"
    )

    # set headings + widths
    for col in columns:
        tree.heading(col, text=col)
        tree.column(col, width=100)

    tree.pack(fill="both", expand=True, padx=10, pady=10)

    # button frame
    btn_frame = tk.Frame(window)
    btn_frame.pack(pady=5)

    # delete selected button (red)
    btn_delete = tk.Button(
        btn_frame,
        text="Delete Selected",
        bg="#E74C3C",
        fg="white"
    )
    btn_delete.pack(side="left", padx=5)

    # clear history button (orange)
    btn_clear = tk.Button(
        btn_frame,
        text="Clear History",
        bg="#F39C12",
        fg="white"
    )
    btn_clear.pack(side="left", padx=5)

    # close button
    btn_close = tk.Button(
        btn_frame,
        text="Close",
        command=window.destroy
    )
    btn_close.pack(side="left", padx=5)

    return {
        "window": window,
        "tree_history": tree,
        "btn_delete": btn_delete,
        "btn_clear": btn_clear,
        "btn_close": btn_close
    }

# open all patient history window
def open_all_patient_history_window(parent):

    history_ui = build_patient_history_window(parent)

    # fetch history rows
    rows = get_all_patient_history()

    # insert each row into table
    for row in rows:
        history_ui["tree_history"].insert(
            "",
            tk.END,
            values=(
                row["id"],
                row["name"],
                row["phone"],
                row["age"],
                row["payment_status"],
                row["visit_date"]
            )
        )

    # delete selected history item
    def delete_selected():
        selected = history_ui["tree_history"].selection()

        # warn + stop if no selection
        if not selected:
            messagebox.showwarning(
                "No Selection",
                "Select a history item first."
            )
            return

        item = selected[0]
        history_id = history_ui["tree_history"].item(item)["values"][0]

        # delete from database + table
        delete_history_item(history_id)
        history_ui["tree_history"].delete(item)

    # clear all history
    def clear_history():
        clear_all_history()

        # remove all rows from table
        for item in history_ui["tree_history"].get_children():
            history_ui["tree_history"].delete(item)

    # connect buttons
    history_ui["btn_delete"].config(
        command=delete_selected
    )

    history_ui["btn_clear"].config(
        command=clear_history
    )
```

```python
# common devices list
COMMON_DEVICES = [
    'Nerve stimulation device',
    'nerve stimulation(face)',
    'Infrared Heat Lamp',
    'massage (relax device)',
    'massage (Thrive 717 Handy Massager)',
]

# get common devices list
def get_common_devices():
    return list(COMMON_DEVICES)

# build recovery plan window
def build_recovery_plan_window(parent, patient_name):
    window = tk.Toplevel(parent)
    window.title(f'Doctor Action — Ordered Recovery Plan — {patient_name}')
    window.geometry('1190x780')
    window.minsize(930, 690)
    # title
    tk.Label(window, text=f'Doctor Action: Recovery Plan — {patient_name}',
             font=('Arial', 16, 'bold'), fg=COLOR_PRIMARY).pack(pady=(10, 5))

    # prescribed visits row
    visits = ttk.Frame(window, padding=(16, 2))
    visits.pack(fill=tk.X)
    ttk.Label(visits, text='Number of visits prescribed:',
              font=('Arial', 11, 'bold')).pack(side=tk.LEFT)
    entry_total_visits = ttk.Entry(visits, width=8)
    entry_total_visits.pack(side=tk.LEFT, padx=12)
    ttk.Label(visits, text='This patient\'s visit total appears in nurse queue, history, and Excel.').pack(side=tk.LEFT)

    # quick device preset buttons
    quick = ttk.LabelFrame(window, text='1. Choose a common treatment device — click to ADD it as the next step', padding=8)
    quick.pack(fill=tk.X, padx=15, pady=(8, 4))
    preset_buttons = {}
    # create one button per device
    for i, device in enumerate(get_common_devices(), 1):
        btn = ttk.Button(quick, text=f'{i}. {device}')
        btn.grid(row=(i - 1) // 2, column=(i - 1) % 2, sticky='ew', padx=7, pady=3, ipady=3)
        preset_buttons[device] = btn
    quick.columnconfigure(0, weight=1)
    quick.columnconfigure(1, weight=1)

    # step editor
    editor = ttk.LabelFrame(window, text='2. Configure selected step — doctor notes are visible to the nurse', padding=(12, 6))
    editor.pack(fill=tk.X, padx=15, pady=5)
    # step description input
    ttk.Label(editor, text='Step / activity:').grid(row=0, column=0, sticky='w', pady=4)
    entry_desc = ttk.Entry(editor)
    entry_desc.grid(row=0, column=1, sticky='ew', padx=8)
    # device dropdown
    ttk.Label(editor, text='Device:').grid(row=1, column=0, sticky='w', pady=4)
    combo_device = ttk.Combobox(editor, values=get_common_devices(), state='normal')
    combo_device.grid(row=1, column=1, sticky='ew', padx=8)
    # duration input
    ttk.Label(editor, text='Duration (min):').grid(row=2, column=0, sticky='w', pady=4)
    entry_duration = ttk.Entry(editor, width=12)
    entry_duration.grid(row=2, column=1, sticky='w', padx=8)
    # doctor note input
    ttk.Label(editor, text='Doctor instructions\nfor this step:').grid(row=3, column=0, sticky='nw', pady=5)
    entry_step_note = tk.Text(editor, height=3, wrap='word', font=('Arial', 10))
    entry_step_note.grid(row=3, column=1, sticky='ew', padx=8, pady=4)
    # hint label
    ttk.Label(editor, text='e.g. Apply olive oil; position electrode sticker 5 cm from the marked area.',
              foreground='#637788').grid(row=4, column=1, sticky='w', padx=8)
    editor.columnconfigure(1, weight=1)

    # step control buttons
    controls = ttk.Frame(window)
    controls.pack(pady=7)
    btn_add = ttk.Button(controls, text='Add Custom / Update Selected Step')
    btn_add.pack(side=tk.LEFT, padx=4)
    btn_new = ttk.Button(controls, text='Clear / New Step')
    btn_new.pack(side=tk.LEFT, padx=4)
    btn_move_up = ttk.Button(controls, text='Move Step Up ↑')
    btn_move_up.pack(side=tk.LEFT, padx=4)
    btn_move_down = ttk.Button(controls, text='Move Step Down ↓')
    btn_move_down.pack(side=tk.LEFT, padx=4)
    btn_remove = ttk.Button(controls, text='Remove Step')
    btn_remove.pack(side=tk.LEFT, padx=4)

    # steps table title
    ttk.Label(window, text='3. Ordered recovery plan —  select a step to edit. Each device click adds a NEW numbered step.',
              font=('Arial', 10, 'bold')).pack(anchor='w', padx=16)
    # steps table
    tree = _medical_tree(window,
        ('Step', 'Description', 'Device', 'Duration', 'Doctor Note'),
        (50, 250, 260, 95, 410), height=8)
    # status label
    status_label = ttk.Label(window, text='Quick start: click a device above, then set its duration and instruction, and click Update Selected Step.',
                             foreground='#305972')
    status_label.pack(pady=3)
    # save plan button
    btn_save = tk.Button(window, text='Save Recovery Plan → Nurse Steps',
                         bg='#1D8665', fg='white', font=('Arial', 12, 'bold'))
    btn_save.pack(pady=(7, 12), ipady=5, ipadx=12)
    # return widgets
    return {'window': window, 'entry_total_visits': entry_total_visits,
            'entry_desc': entry_desc, 'combo_device': combo_device,
            'entry_duration': entry_duration, 'entry_step_note': entry_step_note,
            'tree_steps': tree, 'btn_add_step': btn_add, 'btn_new_step': btn_new,
            'btn_move_up': btn_move_up, 'btn_move_down': btn_move_down,
            'btn_remove_step': btn_remove, 'btn_save_plan': btn_save,
            'preset_buttons': preset_buttons, 'status_label': status_label}

# build execute plan window
def build_execute_plan_window(parent, patient_name, patient_id):
    window = tk.Toplevel(parent)
    window.title(f'Nurse Action — Follow Doctor Recovery Steps — {patient_name}')
    window.geometry('1170x770')
    window.minsize(950, 650)
    # title
    tk.Label(window, text=f'Nurse Action: Ordered Recovery Plan — {patient_name}',
             font=('Arial', 16, 'bold'), fg=COLOR_PRIMARY).pack(pady=(12, 5))
    # visit label
    visit_label = ttk.Label(window, font=('Arial', 11, 'bold'))
    visit_label.pack(pady=4)
    # next step banner
    next_step_label = tk.Label(window, text='STEP 1 — Select the next required treatment',
                               font=('Arial', 12, 'bold'), fg='#16496D',
                               bg='#EAF3F7', anchor='w', padx=15, pady=8)
    next_step_label.pack(fill=tk.X, padx=15, pady=4)
    # steps table
    tree = _medical_tree(window,
        ('Step', 'Description', 'Device', 'Duration', 'Doctor Note', 'Status'),
        (55, 235, 230, 88, 380, 85), height=10)
    # doctor note section title
    ttk.Label(window, text='Doctor instruction for the highlighted step:',
              font=('Arial', 11, 'bold')).pack(anchor='w', padx=15, pady=(5, 0))
    # doctor note label
    note = tk.Label(window, text='No additional doctor notes.', justify='left',
        wraplength=1040, fg='#24445A', bg='#EDF5F7',
        font=('Arial', 11), anchor='w', padx=15, pady=13)
    note.pack(fill=tk.X, padx=15, pady=6)
    # mark step button
    btn_mark = tk.Button(window, text='Mark Current Step Done → Next', bg=COLOR_PRIMARY,
                         fg='white', font=('Arial', 11, 'bold'))
    btn_mark.pack(pady=6, ipady=3)
    # nurse note input
    ttk.Label(window, text='Optional nurse notes for this completed visit:').pack(anchor='w', padx=15)
    visit_note = tk.Text(window, height=2, wrap='word')
    visit_note.pack(fill=tk.X, padx=15, pady=4)
    # finish button
    btn_finish = tk.Button(window, text='Complete Visit and Save History',
                           bg='#218C74', fg='white', font=('Arial', 12, 'bold'))
    btn_finish.pack(pady=(6, 12), ipady=4, ipadx=10)
    # return widgets
    return {'window': window, 'tree_steps': tree, 'patient_id': patient_id,
            'visit_label': visit_label, 'next_step_label': next_step_label,
            'instruction_label': note, 'entry_visit_note': visit_note,
            'btn_mark_step': btn_mark, 'btn_complete': btn_finish}
```
